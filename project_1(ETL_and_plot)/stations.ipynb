```python
#load packages
import requests
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

```python
# Get the list of stations
stations_url = "https://opendataapi.dmi.dk/v2/metObs/collections/station/items"
r = requests.get(stations_url, params={"limit": 1000})
stations = r.json()

# Pull into a DataFrame
station_records = [s["properties"] for s in stations["features"]]
stations_df = pd.DataFrame(station_records)
stations_df.head(100)
```
